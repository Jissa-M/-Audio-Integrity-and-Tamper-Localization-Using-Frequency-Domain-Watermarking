```python

import numpy as np
import matplotlib.pyplot as plt

from audio_processing import (
    load_audio,
    compute_stft,
    reconstruct_audio,
    reconstruction_metrics
)

# Load an audio sample
y, sr = load_audio("data/audio.wav")

# Compute STFT and reconstruct audio
S = compute_stft(y)
y_reconstructed = reconstruct_audio(S, len(y))

# Calculate reconstruction quality
rmse, max_error, snr_db = reconstruction_metrics(
    y, y_reconstructed
)

print("Sample rate:", sr)
print("Duration:", len(y) / sr, "seconds")
print("STFT shape:", S.shape)
print("RMSE:", rmse)
print("Maximum error:", max_error)
print("Reconstruction SNR:", snr_db, "dB")

# Plot waveform
time = np.arange(len(y)) / sr

plt.figure(figsize=(12, 3))
plt.plot(time, y)
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.title("Audio Waveform")
plt.tight_layout()
plt.show()

# Plot reconstruction error
error = y - y_reconstructed

plt.figure(figsize=(12, 3))
plt.plot(time, error)
plt.xlabel("Time (seconds)")
plt.ylabel("Error")
plt.title("Reconstruction Error")
plt.tight_layout()
plt.show()
```
