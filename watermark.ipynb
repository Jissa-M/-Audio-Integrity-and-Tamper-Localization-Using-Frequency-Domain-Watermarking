```python
import hashlib
import numpy as np

def generate_watermark(key, frame_index, num_bins):
    seed_text = f"{key}:{frame_index}"
    seed_bytes = hashlib.sha256(seed_text.encode()).digest()
    seed = int.from_bytes(seed_bytes[:8], byteorder="big")

    rng = np.random.default_rng(seed)

    watermark = rng.choice(
        [-1.0, 1.0],
        size=num_bins
    )

    return watermark
```

```python
key = "test_secret_key"

w1 = generate_watermark(key, 0, 20)
w2 = generate_watermark(key, 0, 20)
w3 = generate_watermark(key, 1, 20)

print("Watermark:", w1)
print("Same frame gives same pattern:", np.array_equal(w1, w2))
print("Different frame gives different pattern:", not np.array_equal(w1, w3))
print("Only -1 and +1:", np.all(np.isin(w1, [-1.0, 1.0])))
```
